Problem 1

Take two integers using input() and print their sum. Remember: input() always returns a string, so 
convert with int() before adding. Sample Input: 5 2 Sample Output: 7

```python
a = int(input("Enter The First Number"))
b = int(input("Enter The Second Number"))

print(a + b)
```

Problem - 2

Take an integer N. Print Even if N is even, otherwise print Odd. Use the modulus operator %. Sample 
Input: 17 Sample Output: Odd

```python
n = int(input("Enter Your Number: "))

if n % 2 == 0:
    print("Even")
else:
    print("Odd")
```

Problem - 3

Take an integer marks. Print the grade using this rule, and put the highest check first: A+ if marks >= 80, 
A if marks >= 70, Pass if marks >= 40, otherwise Fail. Sample Input: 85 Sample Output: A+ Sample 
Input 2: 40 Sample Output 2: Pass

```python
result = int(input("Enter Your Number: "))

if result >= 80:
    print("A+")
elif result >= 70:
    print("A")
elif result >= 40:
    print("Pass")
else:
    print("Fail")
```

Problem - 4

Take an integer N. Use a while loop to print numbers from 1 to N, each on a new line. Do not forget to 
increase the counter. Sample Input: 5 Sample Output: 1 2 3 4 5

```python
n = int(input("Enter Your Number: "))

i = 1
while i <= n:
    print(i)
    i += 1
```

Problem - 5

Take an integer N. Use a for loop and range to print all odd numbers from 1 to N. Sample Input: 10 
Sample Output: 1 3 5 7 9

```python
n = int(input("Enter Your Number: "))

for i in range(1, n + 1, 2):
    print(i)
```

Problem - 6

Take an integer N. Print all numbers from 1 to N, but skip multiples of 3 using continue. Sample Input: 
10 Sample Output: 1 2 4 5 7 8 10

```python
n = int(input("Enter Your Number: "))

for i in range(1, n + 1):
    if i % 3 == 0:
        continue
    print(i)
```

Problem - 7

Keep taking integers as input. Add them to a total. Stop when a negative number appears (break). Do 
not add that negative number. Then print the total. Sample Input: 4 8 15 -3 23 Sample Output: 27 

```python
total = 0
numbers = map(int, input().split())

for num in numbers:
    if num < 0:
        break
    total += num
print(total)
```

Problem - 9

Take a string as input. Count how many vowels it has (a, e, i, o, u in both lowercase and 
uppercase) using a loop. Print only the count. Sample Input: Education Sample Output: 5 

```python
text = input("Enter Your String: ")
vowels = "aeiouAEIOU"
count = 0

for char in text:
    if char in vowels:
        count += 1
print(count)
```
