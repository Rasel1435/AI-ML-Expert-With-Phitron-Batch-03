Problem 1

Take two integers using input() and print their sum. Remember: input() always returns a string, so 
convert with int() before adding. Sample Input: 5 2 Sample Output: 7

```python
a = int(input("Enter The First Number"))
b = int(input("Enter The Second Number"))

print(a + b)
```

Problem - 2

Take an integer N. Print Even if N is even, otherwise print Odd. Use the modulus operator %. Sample 
Input: 17 Sample Output: Odd

```python
n = int(input("Enter Your Number: "))

if n % 2 == 0:
    print("Even")
else:
    print("Odd")
```

Problem - 3

Take an integer marks. Print the grade using this rule, and put the highest check first: A+ if marks >= 80, 
A if marks >= 70, Pass if marks >= 40, otherwise Fail. Sample Input: 85 Sample Output: A+ Sample 
Input 2: 40 Sample Output 2: Pass

```python
result = int(input("Enter Your Number: "))

if result >= 80:
    print("A+")
elif result >= 70:
    print("A")
elif result >= 40:
    print("Pass")
else:
    print("Fail")
```

Problem - 4

Take an integer N. Use a while loop to print numbers from 1 to N, each on a new line. Do not forget to 
increase the counter. Sample Input: 5 Sample Output: 1 2 3 4 5

```python
n = int(input("Enter Your Number: "))

i = 1
while i <= n:
    print(i)
    i += 1
```
