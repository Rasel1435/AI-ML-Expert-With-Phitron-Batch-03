**Problem 1: Smart Light**

A smart light turns **ON** if the room is bright enough. 
You are given two integers: 
- brightness 
- threshold 

If brightness >= threshold, print **ON**. Otherwise, print **OFF**. </br> 
Input Format </br> 
Two integers: brightness and threshold </br> 
Output Format </br> 
ON </br> 
or </br> 
OFF </br> 
Sample Input </br> 
75 60 </br> 
Sample Output </br> 
ON </br> 

```python
brightness = int(input("Enter Your Number Of Brightness: "))
threshold = int(input("Enter Your Number Of Threshold: "))

if brightness >= threshold:
    print("ON")
else:
    print("OFF")
```

### Advance Soluation

```python
brightness, threshold = map(int, input().split())

if brightness >= threshold:
    print("ON")
else:
    print("OFF")
```

### Problem 2: Even or Odd

You are given an integer **N**. </br>
Determine whether the number is **Even** or **Odd**. </br>
**Input Format** </br> 
One integer **N** </br>
**Output Format** </br> 
Even </br>
or </br>
Odd </br>
Sample Input </br>
17 </br>
Sample Output </br>
Odd

```python
n = int(input())

if n % 2 == 0:
    print("Even")
else:
    print("Odd")
```

### Problem 3: Print Numbers

You are given an integer N.</br>
Print all numbers from 1 to N, one number per line. </br>
**Input Format**</br>
One integer **N** </br>
**Sample Input** </br>
5 </br>

**Sample Output** </br>

1 </br>
2 </br>
3 </br>
4 </br>
5

```python
n = int(input())

for i in range(1, n + 1):
    print(i)
```

### Problem 4: Count Even Numbers

You are given N integers.</br>
Count how many of them are even. </br>
**Input Format**</br> 
- First line contains N. 
- Second line contains N integers.

**Sample Input**</br>
6 </br>
2 7 8 11 14 5 </br>
**Sample Output**</br>
3 </br>
Explanation: </br>
 The even numbers are 2, 8, 14. 

```python
n = int(input())

numbers = map(int, input().split())
count = 0
for num in numbers:
    if num % 2 == 0:
        count += 1
print(count)
```

### Problem 5: Sum of Odd Numbers

You are given an integer N. </br>
Find the sum of all odd numbers from 1 to N. </br>
**Input** </br>
10 </br>
**Output** </br>
25 </br>
Explanation: </br>
The odd numbers are: </br>
1 + 3 + 5 + 7 + 9 = 25 </br>
Hint: </br>
 You can use continue to skip the even numbers.

```python
n = int(input())

total_sum = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        continue
    total_sum += i
print(total_sum)
```

```python
n = int(input())
print(sum(range(1, n + 1, 2)))
```
