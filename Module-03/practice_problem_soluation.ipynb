### Problem 1: Smart Light

A smart light turns **ON** if the room is bright enough. 
You are given two integers: 
- brightness 
- threshold 

If brightness >= threshold, print **ON**. Otherwise, print **OFF**. </br> 
Input Format </br> 
Two integers: brightness and threshold </br> 
Output Format </br> 
ON </br> 
or </br> 
OFF </br> 
Sample Input </br> 
75 60 </br> 
Sample Output </br> 
ON </br> 

```python
brightness = int(input("Enter Your Number Of Brightness: "))
threshold = int(input("Enter Your Number Of Threshold: "))

if brightness >= threshold:
    print("ON")
else:
    print("OFF")
```

### Advance Soluation

### Problem 4: Count Even Numbers

You are given N integers.</br>
Count how many of them are even. </br>
**Input Format**</br> 
- First line contains N. 
- Second line contains N integers.

**Sample Input**</br>
6 </br>
2 7 8 11 14 5 </br>
**Sample Output**</br>
3 </br>
Explanation: </br>
 The even numbers are 2, 8, 14. 

```python
n = int(input())

numbers = map(int, input().split())
count = 0
for num in numbers:
    if num % 2 == 0:
        count += 1
print(count)
```

### Problem 5: Sum of Odd Numbers

You are given an integer N. </br>
Find the sum of all odd numbers from 1 to N. </br>
**Input** </br>
10 </br>
**Output** </br>
25 </br>
Explanation: </br>
The odd numbers are: </br>
1 + 3 + 5 + 7 + 9 = 25 </br>
Hint: </br>
 You can use continue to skip the even numbers.

```python
n = int(input())

total_sum = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        continue
    total_sum += i
print(total_sum)
```

### Problem 6: Smallest and Largest

You are given **N** integers.</br> 
Find the smallest and largest number among them.</br>
Do not use Python's built-in min() or max()</br> functions. </br>

**Input Format**</br>
- First line contains N. 
- Second line contains N integers. 

**Sample Input**

5</br>
12 7 25 3 18 </br>

**Sample Output **

3 25 

```python
n = int(input())
numbers = list(map(int, input().split()))

smallest = numbers[0]
largest = numbers[0]

for i in range(1, n):
  if numbers[i] < smallest:
    smallest = numbers[i]
  if numbers[i] > largest:
    largest = numbers[i]

print(smallest, largest)
```

### Problem 7: Positive, Negative, Even and Odd

**You are given N integers.**</br>
Count: </br>
1.  How many numbers are positive 
2.  How many numbers are negative 
3.  How many numbers are even 
4.  How many numbers are odd 

Zero is neither positive nor negative, but zero is even. </br>
**Input Format**</br> 
First line: N </br>
Second line: N integers </br>
**Sample Input**</br> 
7 </br>
-5 0 8 -2 7 11 4 </br>
**Sample Output **</br>
Positive: 4 </br>
Negative: 2 </br>
Even: 4 </br>
Odd: 3</br>

```python
n = int(input())
numbers = list(map(int, input().split()))

positive = 0
negative = 0
even = 0
odd = 0

for num in numbers:
  if num > 0:
    positive += 1
  elif num < 0:
    negative += 1

  if num % 2 == 0:
    even += 1
  else:
    odd += 1

print(f"Positive: {positive}")
print(f"Negative: {negative}")
print(f"Even: {even}")
print(f"Odd: {odd}")
```

```python
def solve():
  n = int(input())
  numbers = list(map(int, input().split()))

  positive = sum(1 for x in numbers if x > 0)
  negative = sum(1 for x in numbers if x < 0)

  even = sum(1 for x in numbers if x % 2 == 0)
  odd = n - even

  print(f"Positive: {positive}")
  print(f"Negative: {negative}")
  print(f"Even: {even}")
  print(f"Odd: {odd}")


if __name__ == "__main__":
  solve()
```

### Problem 8: Reverse Digits

You are given a positive integer **N**.</br> 
Print its digits in reverse order.</br>
**Input**</br>
12345</br>
**Output**</br>
5 4 3 2 1

**For example:**</br> 
**Input**: 908</br> 
**Output**: 8 0 9 </br>

**Hint**: </br>
 Use:</br>
number % 10</br>
to get the last digit, and</br>
number // 10</br>
to remove the last digit.</br>

```python
n = int(input())

while n > 0:
    digit = n % 10
    print(digit, end="")
    n = n // 10
```

### Problem 9: Find the First Number Greater Than 100

You are given N integers.

Find the first number greater than 100.

As soon as you find the first number greater than 100, stop checking the remaining numbers.

If there is no number greater than 100, print:

Not Found

Input Format

First line: N

Second line: N integers

Sample Input

7

20 45 80 105 150 200 300

Sample Output

105

Important:

Even though 150, 200, 300 are also greater than 100, only 105 should be printed because it is the
first one.

```python
n = int(input())
numbers = map(int, input().split())

for i in numbers:
    if i > 100:
        print(i)
        break
else:
    print("Not Found")
```

```python
def find_first_greater_than_100():
    try:
        n = int(input().strip())
        numbers = list(map(int, input().split()))

        # Generator expression with next() for concise O(N) evaluation
        result = next((num for num in numbers if num > 100), "Not Found")
        print(result)

    except (ValueError, EOFError):
        return


if __name__ == "__main__":
    find_first_greater_than_100()
```

### Problem 10: Number Analysis

You are given N positive integers.</br>
For each number:</br>
-    If the number is even, ignore it and move to the next number.
-    If the number is odd, calculate the sum of its digits.
-    If the digit sum is greater than 20, stop processing any more numbers.
-    Otherwise, print the number and its digit sum.

At the end, print how many odd numbers were successfully processed.
Sample Input</br>
6</br>
12 135 47 91 999 25</br>
So the output will be:</br>
135 9</br>
47 11</br>
91 10</br>
3</br>

```python
def analyze_numbers():
    try:
        n = int(input().strip())
        numbers = list(map(int, input().split()))

        processed_count = 0

        for num in numbers:
            # Ignore even numbers
            if num % 2 == 0:
                continue

            # Calculate the sum of its digits
            digit_sum = sum(int(digit) for digit in str(num))

            # Stop processing if digit sum is greater than 20
            if digit_sum > 20:
                break

            # Print the number and its digit sum
            print(f"{num} {digit_sum}")
            processed_count += 1

        # Print the total successfully processed odd numbers
        print(processed_count)

    except (ValueError, EOFError):
        return


if __name__ == "__main__":
    analyze_numbers()
```
