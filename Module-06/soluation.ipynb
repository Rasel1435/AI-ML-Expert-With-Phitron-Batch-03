### A. The Homework Bot (Easy)

**Problem statement**</br> 
The lab built a homework bot. It does not give long speeches. It only answers one question: did </br>
this student pass the AI quiz? </br>
Write a function check_pass(marks) that returns True if marks >= 40, otherwise False. Do not</br>
print inside the function. Add a short docstring — even bots deserve documentation. </br>
Then read one quiz mark, call the function, and print Pass or Fail. </br>
**Input**</br> 
One line: an integer marks (−100 ≤ marks ≤ 100). </br>
**Output*** </br>
Print Pass or Fail. </br>
**Examples**</br> 
**Input**</br> 
73 </br>
**Output**</br> 
Pass</br> 
**Input***</br> 
39 </br>
**Output**</br> 
Fail

```python
def check_pass(marks):
    """Check if the student passed the AI quiz (marks >= 40)."""
    return marks >= 40

quiz_mark = int(input().strip())
print(f"Student Mark: {quiz_mark}")

if check_pass(quiz_mark):
    print("Pass")
else:
    print("Fail")
```

### B. Three Training Runs (Easy)

**Problem statement**</br> 
You trained a tiny model three times. Same data, three different random seeds. The accuracies </br>
came out different — models have moods, like you before coffee. </br>
Write average(a, b, c) that returns the mean accuracy as a float: (a + b + c) / 3. Call it with </br>
**positional** arguments and print only the returned value. The report needs a decimal, so the </br>
supervisor believes you actually divided. </br>
**Input** </br>
One line: three numbers a b c (the three accuracies). </br>
**Output** </br>
Print the average as a float (for example 20.0, not 20). </br>
**Example**</br> 
**Input** </br>
10 20 30 </br>
**Output**</br> 
20.0</br>

```python
a, b, c = map(float, input().split())
result = (a + b + c) / 3
print(result)
```

```python
def average(a, b, c):
    """Returns the mean accuracy as a float."""
    return (a + b + c) / 3

a, b, c = map(float, input().split())
print(average(a, b, c))
```
