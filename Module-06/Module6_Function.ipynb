## Different type of argument

```python
def adder(a,b,c):
    print(f"{a},{b},{c}")
    
adder(10,30,20)
```

```python
def adder(*args):
    for i in args:
        print(i)
        
adder(10,30,20,40)
```

```python
def adder(a=10,b=30,c=10):
    print(f"{a},{b},{c}")
    
adder()
```

```python
def adder(**kwargs):
  for key,value in kwargs.items():
    print(f"{key}: {value}")

adder(a = 10 , b=30 , c = 40 , n_estimator=50)
```

## Return type

```python
def welcome_message():
  print("hello user")

print(type(welcome_message()))
```

```python
def adder(a,b) :
  sum = a+b
  return sum

print(type(adder(10,20)))
```

```python
def adder(a,b) :
  sum = a+b
  return sum

print(type(adder(10.5,20)))
```

```python
def welcome_message(user="None"):
  return f"hello {user}"

print(type(welcome_message()))
```

## iterator

```python
s = {10 , 20 , 30 , 40 , 50 }

s_iter = iter(s)
print(next(s_iter))
print(next(s_iter))
print(next(s_iter))
print(next(s_iter))
print(next(s_iter))

for i in s:
    print(i)
```

## generator

```python
lst = [x for x in range(500)] #dataset

def data_loader(chunk_size, lst):
    for i in range(0,len(lst), chunk_size):
        yield lst[i:i+chunk_size]
        
x = data_loader(5,lst)

print(next(x))
print(next(x))
```
