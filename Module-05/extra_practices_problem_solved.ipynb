Given a string S. Determine how many times does each letter occurred in S.

**Input** </br>
Only one line contains the string S (1 ≤ |S| ≤ 107) where |S| is the length of the string and it consists of only **lowercase** English letters.

**Output** </br>
For each character that appears in S, print a single line that contains the following format: "X : Y" where X is the letter and Y is the number of times that letter X occurred in S.

**Note**: you must print letters in **ascending** order.

**Examples**</br>

**Input**</br>
aaabbc </br>

**Ouput**</br>
a : 3</br>
b : 2</br>
c : 1

**Input**</br>
regff </br>

**Ouput**</br>
e : 1 </br>
f : 2 </br>
g : 1 </br>
r : 1


```python
import sys

def solve() -> None:
    data = sys.stdin.buffer.read()
    if not data:
        return

    freq = [0] * 26
    a_ord = ord("a")
    z_ord = ord("z")

    for byte in data:
        if a_ord <= byte <= z_ord:
            freq[byte - a_ord] += 1

    out = []
    for i in range(26):
        if freq[i] > 0:
            out.append(f"{chr(a_ord + i)} : {freq[i]}")

    if out:
        sys.stdout.write("\n".join(out) + "\n")

if __name__ == "__main__":
    solve()
```

```python
def solve(text: str) -> None:
    if not text:
        return

    data = text.encode("utf-8")
    freq = [0] * 26
    a_ord = ord("a")
    z_ord = ord("z")

    for byte in data:
        if a_ord <= byte <= z_ord:
            freq[byte - a_ord] += 1

    out = [f"{chr(a_ord + i)} : {freq[i]}" for i in range(26) if freq[i] > 0]

    if out:
        print("\n".join(out))


# Change this test string to whatever you want to count
sample_input = "aaabbc"
solve(sample_input)
```
